```python
from PIL import Image, ImageDraw, ImageFont
import textwrap

# Create a large canvas
width = 1600
height = 1200
background_color = (255, 255, 255)
text_color = (0, 0, 0)
box_color = (200, 220, 255)
box_border = (0, 102, 204)
line_color = (100, 100, 100)

img = Image.new('RGB', (width, height), background_color)
draw = ImageDraw.Draw(img)

# Try to use a better font
try:
    title_font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", 18)
    entity_font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", 12)
    attr_font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", 10)
except:
    title_font = ImageFont.load_default()
    entity_font = ImageFont.load_default()
    attr_font = ImageFont.load_default()

# Title
draw.text((width//2 - 150, 20), "PARK DATABASE - ENTITY RELATIONSHIP DIAGRAM", fill=text_color, font=title_font)

# Define entities with their attributes
entities = {
    "LOCATION": {
        "pos": (50, 100),
        "attrs": ["LocationID (PK)", "Street", "City", "Description", "MaxCapacity"]
    },
    "AREA": {
        "pos": (400, 100),
        "attrs": ["AreaID (PK)", "Type", "Description", "Size"]
    },
    "SITE": {
        "pos": (750, 100),
        "attrs": ["AreaID (FK)", "SiteNumber (PK)"]
    },
    "EVENT": {
        "pos": (1100, 100),
        "attrs": ["EventID (PK)", "EventName", "Description"]
    },
    "SPORTING-EVENT": {
        "pos": (50, 350),
        "attrs": ["EventID (PK,FK)", "TeamName"]
    },
    "PERFORMANCE": {
        "pos": (350, 350),
        "attrs": ["EventID (PK,FK)", "Duration"]
    },
    "CONFERENCE": {
        "pos": (650, 350),
        "attrs": ["EventID (PK,FK)", "SponsoringOrg"]
    },
    "PERFORMER": {
        "pos": (950, 350),
        "attrs": ["PerformerID (PK)", "PerformerName", "EventID (FK)"]
    },
    "EQUIPMENT": {
        "pos": (50, 600),
        "attrs": ["EquipmentID (PK)", "Category", "IsRental"]
    },
    "BIKE": {
        "pos": (350, 600),
        "attrs": ["BikeID (PK,FK)", "Color"]
    },
    "BOAT": {
        "pos": (650, 600),
        "attrs": ["BoatID (PK,FK)", "Seats"]
    },
    "SPORTS-EQUIPMENT": {
        "pos": (950, 600),
        "attrs": ["EquipmentID (PK,FK)", "EquipmentType"]
    },
    "RENTAL": {
        "pos": (50, 850),
        "attrs": ["RentalID (PK)", "EquipmentID (FK)", "VisitorID (FK)", "RentalDateTime"]
    },
    "VISITOR": {
        "pos": (500, 850),
        "attrs": ["VisitorID (PK)", "FirstName", "MiddleName", "LastName", "DOB", "RegDate"]
    },
    "VISIT": {
        "pos": (950, 850),
        "attrs": ["VisitID (PK)", "VisitorID (FK)", "LocationID (FK)", "VisitDateTime"]
    }
}

# Draw entities
def draw_entity(entity_name, pos, attrs):
    x, y = pos
    width_box = 220
    height_box = 30 + (len(attrs) * 18)
    
    # Draw box
    draw.rectangle([x, y, x + width_box, y + height_box], fill=box_color, outline=box_border, width=2)
    
    # Draw entity name
    draw.text((x + 5, y + 5), entity_name, fill=text_color, font=entity_font)
    
    # Draw separator line
    draw.line([(x, y + 30), (x + width_box, y + 30)], fill=box_border, width=1)
    
    # Draw attributes
    for i, attr in enumerate(attrs):
        draw.text((x + 5, y + 35 + i * 18), attr, fill=text_color, font=attr_font)
    
    return (x + width_box // 2, y + height_box // 2)

# Store entity centers for relationship lines
entity_centers = {}
for name, data in entities.items():
    entity_centers[name] = draw_entity(name, data["pos"], data["attrs"])

# Draw relationships (simplified representation)
relationships = [
    ("LOCATION", "AREA", "1:N | managed by, contains"),
    ("AREA", "SITE", "1:N | has"),
    ("AREA", "EVENT", "M:N | hosts"),
    ("EVENT", "SPORTING-EVENT", "1:1 | is"),
    ("EVENT", "PERFORMANCE", "1:1 | is"),
    ("EVENT", "CONFERENCE", "1:1 | is"),
    ("PERFORMANCE", "PERFORMER", "1:N | has"),
    ("EQUIPMENT", "RENTAL", "1:N | rented in"),
    ("VISITOR", "RENTAL", "1:N | makes"),
    ("VISITOR", "VISIT", "1:N | makes"),
    ("LOCATION", "VISIT", "1:N | visited in"),
]

# Draw relationship labels
for src, dst, label in relationships:
    if src in entity_centers and dst in entity_centers:
        x1, y1 = entity_centers[src]
        x2, y2 = entity_centers[dst]
        
        # Draw line
        draw.line([(x1, y1), (x2, y2)], fill=line_color, width=2)
        
        # Draw midpoint label
        mid_x = (x1 + x2) // 2
        mid_y = (y1 + y2) // 2
        draw.text((mid_x - 40, mid_y), label, fill=(200, 0, 0), font=attr_font)

# Add legend
legend_y = 1050
draw.text((50, legend_y), "LEGEND:", fill=text_color, font=entity_font)
draw.text((50, legend_y + 30), "PK = Primary Key", fill=text_color, font=attr_font)
draw.text((50, legend_y + 50), "FK = Foreign Key", fill=text_color, font=attr_font)
draw.text((300, legend_y + 30), "1:N = One-to-Many (max cardinality)", fill=text_color, font=attr_font)
draw.text((300, legend_y + 50), "M:N = Many-to-Many (max cardinality)", fill=text_color, font=attr_font)

img.save('park_erd.png')
print("ERD generated successfully as 'park_erd.png'!")
```
